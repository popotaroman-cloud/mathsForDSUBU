## Part 2: Vector Operations — Addition, Scalar Multiply, Dot Product

**Part นี้เราจะทำ Vector operations ทั้งหมดและ verify ผลด้วยตัวเลข เพื่อให้เชื่อมโยงสูตรคณิตศาสตร์กับ Python code ได้**

ใน Machine Learning: การ update น้ำหนัก (weight update) ใน Gradient Descent คือ:
$$\mathbf{w}_{\text{new}} = \mathbf{w}_{\text{old}} - \alpha \cdot \nabla L$$
ซึ่งก็คือ vector subtraction + scalar multiplication นั่นเอง!


````python
# ─── Demo: Vector Operations ──────────────────────────────────────
# วัตถุประสงค์: แสดง vector addition, scalar multiply, dot product
# เพื่อยืนยันว่า NumPy ทำงานตรงกับสูตรคณิตศาสตร์

u = np.array([2.0, 3.0, 1.0])
v = np.array([1.0, -1.0, 4.0])

print("=== Vector Addition ===")
print(f"u = {u}")
print(f"v = {v}")
print(f"u + v = {u + v}")
print(f"u - v = {u - v}")

print("\n=== Scalar Multiplication ===")
print(f"3 * u = {3 * u}")
print(f"0.5 * v = {0.5 * v}")
print(f"Linear combination 2u - v = {2*u - v}")

print("\n=== Dot Product ===")
dot_uv = np.dot(u, v)
print(f"u · v = {dot_uv}")
print(f"Manual check: {u[0]*v[0]} + {u[1]*v[1]} + {u[2]*v[2]} = {u[0]*v[0]+u[1]*v[1]+u[2]*v[2]}")

print("\n=== Norm ===")
norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)
print(f"||u|| = {norm_u:.4f}")
print(f"||v|| = {norm_v:.4f}")

print("\n=== Cosine Similarity ===")
cos_sim = dot_uv / (norm_u * norm_v)
print(f"cos(u, v) = {cos_sim:.4f}")
import math
angle = math.acos(cos_sim) * 180 / math.pi
print(f"angle = {angle:.2f} degrees")
````

## Part 3: Visualization — วาด Vectors ใน 2D

**Part นี้เราจะ visualize vectors ด้วย matplotlib เพื่อให้เห็นภาพเรขาคณิตของ vector operations — การ visualize ช่วยให้เข้าใจและจำนิยามได้ดีกว่าการดูสูตรอย่างเดียว**

การ visualize ข้อมูลเป็นทักษะสำคัญสำหรับ Data Scientist — ก่อนที่จะ build model ควร plot ข้อมูลเสมอเพื่อเข้าใจ structure ของ data


````python
# ─── Demo: วาด Vectors ใน 2D ──────────────────────────────────────
# วัตถุประสงค์: แสดงการ visualize vector addition และ scalar multiplication
# เพื่อให้เข้าใจการตีความเรขาคณิตของ vector operations

import numpy as np
import matplotlib.pyplot as plt

def draw_vector(ax, vec, origin=np.array([0,0]), color='blue', label=''):
    """วาด vector ด้วยลูกศร จาก origin ไปยัง vec"""
    ax.annotate('', xy=origin+vec, xytext=origin,
                arrowprops=dict(arrowstyle='->', color=color, lw=2))
    if label:
        ax.text(origin[0]+vec[0]/2+0.1, origin[1]+vec[1]/2+0.1,
                label, color=color, fontsize=12, fontweight='bold')

# สร้าง vectors
u = np.array([2.0, 1.0])
v = np.array([1.0, 2.0])

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# ── Plot 1: Vector Addition ──
ax = axes[0]
draw_vector(ax, u, color='blue', label='u=(2,1)')
draw_vector(ax, v, color='red', label='v=(1,2)')
draw_vector(ax, u+v, color='green', label='u+v=(3,3)')
# แสดง parallelogram rule
draw_vector(ax, v, origin=u, color='red')
draw_vector(ax, u, origin=v, color='blue')

ax.set_xlim(-0.5, 4)
ax.set_ylim(-0.5, 4)
ax.axhline(0, color='k', lw=0.5)
ax.axvline(0, color='k', lw=0.5)
ax.grid(True, alpha=0.3)
ax.set_title('Vector Addition (Parallelogram Rule)')
ax.set_xlabel('x₁')
ax.set_ylabel('x₂')

# ── Plot 2: Scalar Multiplication ──
ax = axes[1]
draw_vector(ax, u, color='blue', label='u=(2,1)')
draw_vector(ax, 2*u, color='purple', label='2u=(4,2)')
draw_vector(ax, -1*u, color='orange', label='-u=(-2,-1)')
draw_vector(ax, 0.5*u, color='cyan', label='0.5u=(1,0.5)')

ax.set_xlim(-3, 5)
ax.set_ylim(-2, 3)
ax.axhline(0, color='k', lw=0.5)
ax.axvline(0, color='k', lw=0.5)
ax.grid(True, alpha=0.3)
ax.set_title('Scalar Multiplication (c × u)')
ax.set_xlabel('x₁')
ax.set_ylabel('x₂')

plt.tight_layout()
plt.savefig('vector_operations.png', dpi=100, bbox_inches='tight')
plt.show()
print("บันทึก vector_operations.png แล้ว")
````
